```python
test_names = 'John Doe'
```

```python
def parse_and_display(name):
    for i in range(len(name)):
        if name[i] == ' ':
            first = name[:i]
            last = name[i+1:]
            print (f"Full name:{name}")
            print (f"First name: {first}")
            print (f"Last name: {last}")
parse_and_display("John Smith")
```

```python
test_list = ['Lebron James', 'Kevin Durrant', 'Kobe Bryant', 'Jr. Smith']
```

```python
parse_and_display(test_list)
for name in test_list:
    parse_and_display(name)
```
