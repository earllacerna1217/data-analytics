# Today

```python
import datetime
today = datetime.datetime.now()

print(f"Today is {today.strftime('%B %d, %Y')}")
```

# My birthday

```python
birthday = datetime.date(2004,12,17)
print("My birthday is", birthday.strftime("%m/%d/%Y"))
```

# Next 90 Days

```python
import datetime
today = datetime.datetime.now()

ninety_d = today + datetime.timedelta(days=90)
```

```python
print("90 days from now is",  ninety_d.strftime("%B %d, %Y"))
```

# Time for Dinner

```python
dinner_time = datetime.time(19,45)
print("Let's meet for dinner at", dinner_time.strftime("%I:%M %p"))
```
