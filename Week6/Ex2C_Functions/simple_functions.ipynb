# Lab 1 of Exercise 2.C, working with defining custom functions. 

```python
def favorite_things(name, movie, music):
    return f"{name}, {movie}, {music}"
name = "Earl"
movie = 'Real steel'
music = "Body bag"
    
print(favorite_things(name, movie, music))
```

```python
def why_im_here(purpose):
    return f"{purpose}"
purpose = "Aquire the skills to be a Data Analyst"

print(why_im_here(purpose))
```

```python
def favorite_place (place, why):
    return f"{place}, {why}"
place = "Puerto Rico"
why = "Beautiful Island"
print(favorite_place(place, why))
```

```python
place2 = "Philippines"
why2 = "Childhood home"

print(favorite_place(place2, why2))
      
```
